# Capitulo 8 - Funcoes

```python
# Definindo uma funcao simples

def cumprimentar(nome):
    print(f'Ola, {nome.title()}!')

cumprimentar('sara')
```

## Passando argumentos

```python
# Argumentos posicionais (a ordem importa)

def descrever_bicho(tipo_animal, nome_bicho):
    print(f'Eu tenho um {tipo_animal} chamado {nome_bicho.title()}.')

descrever_bicho('cachorro', 'rex')
```

```python
# Argumentos nomeados (a ordem nao importa)

descrever_bicho(nome_bicho='willie', tipo_animal='cachorro')
```

```python
# Valor default para um parametro

def descrever_bicho(nome_bicho, tipo_animal='cachorro'):
    print(f'Eu tenho um {tipo_animal} chamado {nome_bicho.title()}.')

descrever_bicho('rex')
descrever_bicho('whiskers', tipo_animal='gato')
```

## Valores de retorno

```python
# Retornando um valor simples

def nome_completo(primeiro, ultimo):
    nome = f'{primeiro} {ultimo}'
    return nome.title()

musico = nome_completo('jimi', 'hendrix')
print(musico)
```

```python
# Retornando um dicionario

def montar_pessoa(primeiro, ultimo):
    pessoa = {'primeiro_nome': primeiro, 'ultimo_nome': ultimo}
    return pessoa

pessoa = montar_pessoa('jimi', 'hendrix')
print(pessoa)
```

## Passando uma lista

```python
def cumprimentar_usuarios(nomes):
    for nome in nomes:
        print(f'Ola, {nome.title()}!')

nomes_usuarios = ['hannah', 'ty', 'margot']
cumprimentar_usuarios(nomes_usuarios)
```

## Numero arbitrario de argumentos

```python
# O * antes do nome deixa a funcao aceitar quantos ingredientes quiser

def fazer_pizza(*ingredientes):
    print('\nFazendo uma pizza com os ingredientes:')
    for ingrediente in ingredientes:
        print(f'- {ingrediente}')

fazer_pizza('pepperoni')
fazer_pizza('cogumelos', 'pimentao verde', 'queijo extra')
```

```python
# ** antes do nome deixa a funcao aceitar pares chave-valor extras

def montar_perfil(primeiro, ultimo, **informacoes):
    perfil = {'primeiro_nome': primeiro, 'ultimo_nome': ultimo}
    perfil.update(informacoes)
    return perfil

perfil_usuario = montar_perfil('albert', 'einstein', cidade='ulm', campo='fisica')
print(perfil_usuario)
```
