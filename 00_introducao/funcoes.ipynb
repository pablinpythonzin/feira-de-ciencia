# Capitulo 8 - Funcoes

```python
# Definindo uma funcao simples

def cumprimentar(nome):
    print(f'Ola, {nome.title()}!')

cumprimentar('sara')
```

## Passando argumentos

```python
# Argumentos posicionais (a ordem importa)

def descrever_bicho(tipo_animal, nome_bicho):
    print(f'Eu tenho um {tipo_animal} chamado {nome_bicho.title()}.')

descrever_bicho('cachorro', 'rex')
```

```python
# Argumentos nomeados (a ordem nao importa)

descrever_bicho(nome_bicho='willie', tipo_animal='cachorro')
```

```python
# Valor default para um parametro

def descrever_bicho(nome_bicho, tipo_animal='cachorro'):
    print(f'Eu tenho um {tipo_animal} chamado {nome_bicho.title()}.')

descrever_bicho('rex')
descrever_bicho('whiskers', tipo_animal='gato')
```

## Valores de retorno

```python
# Retornando um valor simples

def nome_completo(primeiro, ultimo):
    nome = f'{primeiro} {ultimo}'
    return nome.title()

musico = nome_completo('jimi', 'hendrix')
print(musico)
```

```python
# Retornando um dicionario

def montar_pessoa(primeiro, ultimo):
    pessoa = {'primeiro_nome': primeiro, 'ultimo_nome': ultimo}
    return pessoa

pessoa = montar_pessoa('jimi', 'hendrix')
print(pessoa)
```
