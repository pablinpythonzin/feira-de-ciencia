```python
# Célula 1: Adicionando e alterando elementos em uma lista (Capítulo 3)

# Criando a lista inicial de carros
carros = ["Fusca", "Civic", "Corolla"]
print(f"Lista original: {carros}")

#Modificando um elemento existente pelo índice
carros[0] = "Golf"
print(f"Após modificar o índice 0: {carros}")

#Adicionando um elemento ao final da lista com .append()
carros.append("Mustang")
print(f"Após usar .append(): {carros}")

#Inserindo um elemento em uma posição específica com .insert(índice, valor)
carros.insert(1, "Porsche")
print(f"Após usar .insert() na posição 1: {carros}")
```

```python
# Célula 2: Formas de remover elementos de uma lista (Capítulo 3)

motos = ["Honda", "Yamaha", "Suzuki", "Ducati", "BMW"]

# 1. Usando instrução 'del' (remove permanentemente pelo índice)
del motos[0]
print("Após 'del motos[0]':", motos)

# 3. Usando .remove() (remove pelo valor do texto)
motos.remove("Suzuki")
print("Após .remove('Suzuki'):", motos)
```
