```python
# Um dicionario simples

alien_0 = {'cor': 'verde', 'pontos': 5}
print(alien_0['cor'])
print(alien_0['pontos'])
```

## Trabalhando com dicionarios

```python
# Adicionando novos pares chave-valor

alien_0 = {'cor': 'verde', 'pontos': 5}
alien_0['x_pos'] = 0
alien_0['y_pos'] = 25
print(alien_0)
```

```python
# Comecando com um dicionario vazio

alien_1 = {}
alien_1['cor'] = 'amarelo'
alien_1['pontos'] = 10
print(alien_1)
```

```python
# Modificando valores em um dicionario

alien_0 = {'cor': 'verde', 'pontos': 5}
print(f'A cor original era: {alien_0["cor"]}')

alien_0['cor'] = 'amarelo'
print(f'A cor agora e: {alien_0["cor"]}')
```

```python
# Removendo um par chave-valor com del

alien_0 = {'cor': 'verde', 'pontos': 5}
del alien_0['pontos']
print(alien_0)
```

```python
# Usando get() para acessar um valor com seguranca

alien_0 = {'cor': 'verde'}
pontos = alien_0.get('pontos', 'Nenhuma pontuacao atribuida.')
print(pontos)
```

## Percorrendo um dicionario com um loop

```python
usuario = {
    'nome': 'eric',
    'sobrenome': 'matthes',
    'idade': 25,
}

# percorrendo todos os pares chave-valor
for chave, valor in usuario.items():
    print(f'\nChave: {chave}')
    print(f'Valor: {valor}')
```

```python
linguagens_favoritas = {
    'ana': 'python',
    'bruno': 'c',
    'carla': 'python',
}

# percorrendo so as chaves
for nome in linguagens_favoritas.keys():
    print(nome.title())
```

```python
# percorrendo so os valores
for linguagem in linguagens_favoritas.values():
    print(linguagem.title())
```

## Aninhamento: uma lista de dicionarios

```python
alien_0 = {'cor': 'verde', 'pontos': 5}
alien_1 = {'cor': 'amarelo', 'pontos': 10}
alien_2 = {'cor': 'vermelho', 'pontos': 15}

alienigenas = [alien_0, alien_1, alien_2]

for alien in alienigenas:
    print(alien)
```
