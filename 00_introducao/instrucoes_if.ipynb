# Capitulo 5 - Instrucoes if

```python
# Um exemplo simples

carros = ['audi', 'bmw', 'subaru', 'toyota']
for carro in carros:
    if carro == 'bmw':
        print(carro.upper())
    else:
        print(carro.title())
```

## Testes condicionais

```python
# Verificando a igualdade

carro = 'bmw'
print(carro == 'bmw')
```

```python
# Verificando a diferenca

topping_pedido = 'cogumelos'
if topping_pedido != 'anchovas':
    print('Segue em frente, adicione cogumelos na pizza!')
```

```python
# Comparacoes numericas

idade = 19
print(idade < 21)
print(idade >= 21)
```

```python
# Verificando multiplas condicoes com 'and' e 'or'

idade_0 = 22
idade_1 = 18

print(idade_0 >= 21 and idade_1 >= 21)
print(idade_0 >= 21 or idade_1 >= 21)
```

```python
# Verificando se um valor esta (ou nao) em uma lista

toppings_pedidos = ['cogumelos', 'cebola', 'abacaxi']

print('cogumelos' in toppings_pedidos)
print('pepperoni' in toppings_pedidos)
print('azeitona' not in toppings_pedidos)
```

## Instrucoes if

```python
# if simples

idade = 19
if idade >= 18:
    print('Voce ja pode votar!')
```

```python
# if-else

idade = 17
if idade >= 18:
    print('Voce ja pode votar!')
else:
    print('Desculpe, voce ainda nao tem idade para votar.')
```

```python
# if-elif-else

idade = 12

if idade < 4:
    preco = 0
elif idade < 18:
    preco = 5
else:
    preco = 10

print(f'Voce deve pagar R$ {preco}.')
```

## Usando if com listas

```python
# Verificando se a lista tem itens antes de usa-la

toppings_pedidos = []

if toppings_pedidos:
    for topping in toppings_pedidos:
        print(f'Adicionando {topping}.')
else:
    print('Voce tem certeza de que quer uma pizza sem cobertura?')
```
