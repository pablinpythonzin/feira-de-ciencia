## A funcao input()

As duas celulas abaixo usam `input()`, entao elas so pedem algo digitado quando voce roda o notebook de verdade (por isso nao tem uma saida ja pronta aqui).

```python
nome = input('Qual e o seu nome? ')
print(f'Ola, {nome}!')
```

```python
# Usando int() para transformar a entrada (que sempre vem como texto) em numero
idade = input('Quantos anos voce tem? ')
idade = int(idade)

if idade >= 18:
    print('Voce ja pode votar!')
else:
    print('Voce ainda nao pode votar.')
```

```python
# Operador modulo (%): devolve o resto de uma divisao
print(4 % 3)
print(5 % 3)
print(6 % 3)
```

## Introduzindo os loops while

```python
# Um loop while simples

contador = 1
while contador <= 5:
    print(contador)
    contador += 1
```

```python
# Usando uma flag para controlar quando o loop deve parar

ativo = True
contador = 0

while ativo:
    contador += 1
    print(f'Rodada {contador}')
    if contador == 3:
        ativo = False
```

```python
# Usando break para sair do loop imediatamente

contador = 0
while True:
    contador += 1
    if contador == 3:
        break
    print(contador)
```

```python
# Usando continue para pular para a proxima repeticao

contador = 0
while contador < 10:
    contador += 1
    if contador % 2 == 0:
        continue
    print(contador)
```

```python
idade = int(input("Digite sua idade: "))

if idade <= 11:
    print("Criança")

elif idade <= 18:
    print("jovem")

else:
    print("velho")
```
