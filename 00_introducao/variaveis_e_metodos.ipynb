Conhecendo os metodos basicos

```python
#Usando metodo title()

msg = "todas as primeiras letras serão maiuscula"
print(msg.title())
```

```python
#Usando o strip para remover espaço em brancos

msg = "   Teste de espaço       "
print(f"{msg.strip()}")
```

```python
#usando os metodos Lower() e Upper()

msg = "uma mensagem bem legal"
print(msg.upper())
print(msg.lower())
```
